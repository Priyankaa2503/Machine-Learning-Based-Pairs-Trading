```python
import numpy as np
import pandas as pd
import os
```

```python
df = pd.read_csv(
    "/Users/priyankaramachandran/Desktop/ML Project/dataset/sp500_stocks.csv"
)
print(df.columns.tolist())
print("-" * 80)
```

```python
# sort stocks by date
stocks = df.sort_values("Date")
stocks.head()
```

```python
stocks["Date"] = pd.to_datetime(stocks["Date"])
stocks = stocks.drop_duplicates(subset=["Date", "Symbol"], keep="first")


price_matrix = stocks.pivot(index="Date", columns="Symbol", values="Close").sort_index()
print("Pivoted price matrix shape:", price_matrix.shape)
display(price_matrix.head())


full_dates = pd.date_range(
    start=price_matrix.index.min(), end=price_matrix.index.max(), freq="B"
)
price_matrix = price_matrix.reindex(full_dates)
price_matrix.index.name = "Date"
print("Shape after setting common date range:", price_matrix.shape)

missing_counts = price_matrix.isna().sum().sort_values(ascending=False)
print("Missing values per ticker (top 20):")
display(missing_counts.head(20))
missing_counts.to_csv("missing_counts.csv")

print("Number of tickers with zero NaNs:", (missing_counts == 0).sum())


total_days = price_matrix.shape[0]
missing_percent = (missing_counts / total_days) * 100
valid_tickers = missing_percent[missing_percent <= 5].index
cleaned_prices = price_matrix[valid_tickers]

print(
    f"Kept {len(valid_tickers)} tickers out of {price_matrix.shape[1]} after filtering >5% missing."
)

output_dir = "/Users/priyankaramachandran/Desktop/ML Project/dataset/"
os.makedirs(output_dir, exist_ok=True)

cleaned_prices.to_csv(os.path.join(output_dir, "cleaned_data.csv"), index=True)
print("Cleaned price matrix saved successfully as CSV!")
```
