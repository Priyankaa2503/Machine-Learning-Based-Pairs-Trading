```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("./dataset/cleaned_data.csv", index_col="Date")
```

```python
data.index = pd.to_datetime(data.index)
```

```python
prices = data.copy()
prices = prices.sort_index()
print("Price matrix shape:", prices.shape)
display(prices.head())
```

```python
print(prices.dtypes)
```

```python
print(prices.isna().sum())
```

```python
returns = prices.pct_change()
log_returns = np.log(prices / prices.shift(1))
```

```python
returns.columns = [c + "_ret" for c in returns.columns]
log_returns.columns = [c + "_logret" for c in log_returns.columns]
```

```python
prices.head()
```

```python
rolling_mean = prices.rolling(window=20).mean()
rolling_std = prices.rolling(window=20).std()
```

```python
rolling_mean.columns = [c + "_rollmean20" for c in rolling_mean.columns]
rolling_std.columns = [c + "_rollstd20" for c in rolling_std.columns]
```

```python
rolling_mean.head(5)
```

```python
rolling_std.head()
```

```python
def compute_RSI(series: pd.Series, window=14):
    delta = series.diff()
    gain = delta.clip(lower=0)
    loss = -delta.clip(upper=0)

    avg_gain = gain.rolling(window).mean()
    avg_loss = loss.rolling(window).mean()

    RS = avg_gain / avg_loss
    RSI = 100 - (100 / (1 + RS))
    return RSI


rsi_df = pd.DataFrame(index=prices.index)

for col in prices.columns:
    rsi_df[col + "_RSI14"] = compute_RSI(prices[col], 14)
```

```python
rsi_df.head(20)
```

```python
feature_matrix = pd.concat(
    [
        prices,
        returns,
        log_returns,
        rolling_mean,
        rolling_std,
        rsi_df,
    ],
    axis=1,
)
```

```python
feature_matrix.head()
```

```python
feature_matrix.to_csv("./dataset/features.csv", index=True)
```
